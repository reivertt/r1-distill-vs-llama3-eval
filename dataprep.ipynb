### Odyssey Math Prep

```python
import json
import random

# Configuration
input_path = 'odyssey-math/final-odyssey-math-with-levels.jsonl'
output_path = 'sampled_data/odyssey_sample.jsonl'
SAMPLE_SIZE = 200

entries = []
i = 1
with open(input_path, 'r') as input:
    for line in input:
        data = json.loads(line)
        for _, entry in data.items():
            entry['task_id'] = f"Odyssey/{i}"
            i += 1
            entries.append(entry)

sampled = random.sample(entries, SAMPLE_SIZE)

with open(output_path, 'w') as output:
    for entry in sampled:
        output_json = json.dumps({
            "task_id":entry['task_id'],
            "instruction":entry['question'],
            "input":"",
            "output":"",
            "ground_truth":entry['answer'],
            "reasoning":entry['reasoning'],
            "flag":f"{entry['label']} - {entry['level']}", 
            "deepseek": {
                "is_correct":None,
                "output":""
            },
            "llama" : {
                "is_correct":None,
                "output":""
            }
        })
        output.write(output_json + '\n')

print(f"sampled {len(sampled)} entries and saved to {output_path}")
```

### GSM8K prep

```python
import json
import random

# Configuration
input_path = 'grade-school-math/grade_school_math/data/train_socratic.jsonl'
output_path = 'sampled_data/gsm8k_sample.jsonl'
SAMPLE_SIZE = 100

# Read all entries and parse answers
all_entries = []
i = 1
with open(input_path, 'r') as infile:
    for line in infile:
        data = json.loads(line)
        data['task_id'] = f"GSM8K/{i}"
        i += 1
        # print(data)
        all_entries.append(data)


def parse_ans(x):
    temp = x.split("####")
    res = temp[-1].strip() if len(temp) > 1 else ""
    return res

# Random sampling
sampled_entries = random.sample(all_entries, min(SAMPLE_SIZE, len(all_entries)))

# Write sampled data with extracted answers
with open(output_path, 'w') as outfile:
    for entry in sampled_entries:
        ground_truth = parse_ans(entry['answer'])
        output_json = json.dumps({
            "task_id": entry['task_id'],
            "instruction": entry['question'],
            "input": "",
            "output": "",
            "ground_truth": ground_truth,
            "reasoning": entry['answer'],
            "flag": "",
            "deepseek": {"is_correct": None, "output": ""},
            "llama": {"is_correct": None, "output": ""}
        })
        outfile.write(output_json + '\n')

print(f"Sampled {len(sampled_entries)} entries with extracted answers saved to {output_path}")
```

## Code
### MBPP

```python
import json
import random

# Configuration
input_path = 'sanitized-mbpp.jsonl'
output_path = 'sampled_data/mbpp_sample.jsonl'
all_entries = []
SAMPLE_SIZE = 100

# Function to manipulate prompt
def manipulate_prompt(code: str) -> str:
    first_line = code.strip().split(":")[0]  
    idx1 = first_line.find("def ") + 4    
    idx2 = first_line.find("(")
    idx3 = first_line.find(")") + 1  
    function_ = first_line[idx1:idx2] 
    function_wa = first_line[idx1:idx3]
    
    return function_, function_wa

# Read all entries and parse answers
with open(input_path, 'r') as infile:
    for line in infile:
        data = json.loads(line)
        # print(data)
        all_entries.append(data)

# Random sampling
sampled_entries = random.sample(all_entries, min(SAMPLE_SIZE, len(all_entries)))

# Write sampled data with extracted answers
with open(output_path, 'w') as outfile:
    for entry in sampled_entries:
        function_, function_wa = manipulate_prompt(entry['code'])
        
        tests = "def check(candidate):"
        for i in entry['test_list']:
            tests = f"{tests}\n    {i}"

        output_json = json.dumps({
            "task_id": f"MBPP/{entry['task_id']}",
            "prompt": f"{entry['prompt']} {function_wa}",
            "placeholder": "",
            "test": tests,
            "entry_point": function_,
            "canonical_solution":entry['code']
        })
        outfile.write(output_json + '\n')

print(f"Sampled {len(sampled_entries)} entries with extracted answers saved to {output_path}")
```
