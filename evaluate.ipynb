````python
import json
import random

rob_path = 'sampled_data/big-navigate_sample.jsonl'
sol_path = 'evaluated_data/llama_big-navigate_evaluated.jsonl'
out_path = 'evaluated_data/llama_big-navigate_graded.jsonl'

problems = []
solutions = []

j = 0

pass1 = 0
pass3 = 0
with open(prob_path, 'r') as input1, open(sol_path, 'r') as input2:
    for line in input1:
        problems.append(json.loads(line))

    for line in input2:
        solutions.append(json.loads(line))

    print(f"Loaded {len(problems)} rows from problems and {len(solutions)} rows from solutions")

with open(out_path, 'w') as outfile:
    for problem in problems:
        results = []
        sum_ = 0
        for i in range(3):
            truth = problem["ground_truth"]
            answer = solutions[j]["output"]
            # print(f"{truth} -- {answer}")
            results.append({
                "attempt": i,
                "correct": truth[answer]
            })
            pass3 += truth[answer]
            sum_ += truth[answer]
            j += 1

        output_json = json.dumps({
            "task_id": problem["task_id"],
            "results": results,
            "total":sum_
        })
        outfile.write(output_json + '\n')
        temp = random.choice(results)
        pass1 += temp["correct"]

print(f"===== Llama - BIG Navigate =====")
print(f"pass@1: {pass1/100}")
print(f"pass@3: {pass3/300}")
````

````python
import json
import random
import re

sol_path = 'evaluated_data/deepseek_odyssey_evaluated.jsonl'
problems = []

results = 'evaluated_data/deepseek_odyssey_graded.jsonl'
grades = []

grade_path = 'evaluated_data/deepseek_odyssey_gradedfile.jsonl'

def parse_num(s):
    numbers = re.findall(r'\d+', s)  # Extract all digit sequences
    if not numbers:
        return 0  # Default case if no numbers are found
    num = int(numbers[0])  # Take the first number found
    if '.' in s:
        num = round(float(numbers[0]))  # Handle decimal cases
    return max(0, min(num, 10))  # Ensure the number is between 0 and 10

j = 0

with open(results, 'r') as input2:
    for line in input2:
        data = json.loads(line)
        grade_index = data["output"].find("### Grade")
        if grade_index != -1:
            grade_str = data["output"][grade_index:].split("\n")[1].strip()
            grades.append(parse_num(grade_str))
            # print(f"========== {data["task_id"]} ==========")
            # print(parse_num(grade_str))
        else:
            print(f"========== {data["task_id"]} ==========")
            print(data["output"])
            grades.append(data["output"])

print(len(grades))

with open(sol_path, 'r') as input:
    for line in input:
        problems.append(json.loads(line))

with open(grade_path, 'w') as outfile:
    for problem in problems:
        output_json = json.dumps({
            "task_id":problem["task_id"],
            "grade":grades[j]
        })
        j += 1
        outfile.write(output_json + '\n')
````
