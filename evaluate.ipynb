### LLAMA

````python
import json

# Configuration
prob_path = 'sampled_data/gsm8k_sample.jsonl'
sol_path = 'responses/llama_gsm8k_responses.jsonl'

problems = []
solutions = []
manual_nguli = []

j = 0

with open(prob_path, 'r') as input1, open(sol_path, 'r') as input2:
    for line in input1:
        problems.append(json.loads(line))

    for line in input2:
        solutions.append(json.loads(line))

    print(f"Loaded {len(problems)} rows from problems and {len(solutions)} rows from solutions")

for problem in problems:
    gt = problem["ground_truth"]
    total = 0
    for _ in range(3):
        answers = solutions[j]["output"]
        if answers.find(gt) != -1:
            manual_nguli.append({
                "i":j,
                "id":solutions[j]["task_id"],
                "gt":gt,
                "output":answers[answers.find(gt) - len(gt):]
            })
        else:    
            total += 1
        j += 1
    problem["llama_incorrect"] = total
    print(f"Line {j/3} had {total} incorrect answers")
            
````

````python
for item in manual_nguli:
    print(item)

manual = [
    4,
    5,
    48,
    102,
    103,
    104,
    120,
    123,
    124,
    125,
    135,
    142,
    143,
    148,
    149,
    169,
    231,
    234,
    235,
    236,
    240,
    264,
    265,
    266,
    275,
]
````

````python
j = 0
for problem in problems:
    additional = 0
    for _ in range(3):
        for i in manual:
            if i == j:
                additional +=1
        j += 1
    problem["llama_incorrect"] += additional


acc = 0
with open('evaluated_data/llama_gsm8k_evaluated.jsonl', 'w') as output:
    for problem in problems:
        acc += (3 - problem['llama_incorrect'])
        output_json = json.dumps({
            "task_id": problem['task_id'],
            "correct": 3 - problem['llama_incorrect']
        })
        output.write(output_json + '\n')

print(acc/300)
````
